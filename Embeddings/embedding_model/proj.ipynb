```python
from sentence_transformers import SentenceTransformer
import spacy 
from langchain_text_splitters import RecursiveCharacterTextSplitter # document to chunks
import faiss
```

```python
nlp=spacy.load('en_core_web_sm')
```

```python
with open("data.txt", encoding="utf-8") as f:
    data = f.read()
```

```python
# small data set so 100
splitter= RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=30
)
```

```python
chunks = splitter.split_text(data)

print(chunks[:5])
print(type(chunks))
print(type(chunks[0]))
print("Total Chunks :", len(chunks))
```

```python
# chunks=splitter.split_text(data)

# print(chunks,type(chunks),type(chunks)[0])
```

```python
embeddings = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
# embeddings=SentenceTransformer(
#     model_name_or_path="sentence-transformers/all-miniLM-L6-V2"
# )
```

```python
embedding_vectors = embeddings.encode(
    chunks,
    convert_to_numpy=True
).astype("float32")

print(embedding_vectors.shape)
```

```python
dimension = embedding_vectors.shape[1]

print(dimension)
```

```python
faiss.normalize_L2(embedding_vectors)
```
